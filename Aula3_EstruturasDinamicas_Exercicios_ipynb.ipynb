## Atividade 1 — Listas
Crie um programa que:
1. crie uma lista com 10 números inteiros;
2. solicite os valores ao usuário;
3. exiba todos os números;
4. calcule a soma;
5. calcule a média.
**Desafio:** Encontre o maior valor, o menor valor e a quantidade de números pares.

```python
# Criação da lista vazia
valores = []

print("=== Entrada de Dados ===")
# Leitura dos 10 valores
for contador in range(1, 11):
    numero = int(input(f"Informe o {contador}º valor inteiro: "))
    valores.append(numero)

# Exibição da lista
print("\nValores armazenados:", valores)

# Cálculos básicos usando funções nativas
total = sum(valores)
tamanho_lista = len(valores)
media = total / tamanho_lista

print(f"Somatório: {total}")
print(f"Média calculada: {media:.2f}")

# --- RESOLUÇÃO DO DESAFIO ---
print(f"\nO maior número da lista é: {max(valores)}")
print(f"O menor número da lista é: {min(valores)}")

# Lógica para contar pares
pares = 0
for v in valores:
    if v % 2 == 0:
        pares += 1

print(f"Total de números pares encontrados: {pares}")
```

## Atividade 2 — Matrizes
Crie uma matriz 3 × 3. Solicite os valores ao usuário e:
1. exiba a matriz;
2. calcule a soma de todos os elementos;
3. calcule a soma da diagonal principal;
4. encontre o maior elemento.

```python
grade = []

print("=== Preenchendo a Matriz 3x3 ===")
# Lendo os dados para formar a matriz
for l in range(3):
    linha_atual = []
    for c in range(3):
        item = int(input(f"Insira o valor para a coordenada [{l}][{c}]: "))
        linha_atual.append(item)
    grade.append(linha_atual)

# 1. Exibição formatada
print("\n--- Visualização da Matriz ---")
for l in range(3):
    for c in range(3):
        print(f"[{grade[l][c]:^3}]", end=" ")
    print()

# 2. Soma de todos os elementos
soma_geral = 0
for linha_atual in grade:
    soma_geral += sum(linha_atual)
print(f"\nSoma total da matriz: {soma_geral}")

# 3. Soma da diagonal principal
soma_diag = 0
for indice in range(3):
    soma_diag += grade[indice][indice]
print(f"Soma da diagonal principal: {soma_diag}")

# 4. Encontrando o maior elemento
maior_valor = grade[0][0]
for linha_atual in grade:
    for item in linha_atual:
        if item > maior_valor:
            maior_valor = item

print(f"O maior valor presente na matriz é: {maior_valor}")
```

## Atividade 3 — Dicionário
Crie um dicionário para representar um livro (titulo, autor, ano, preco). Depois:
1. leia os dados;
2. armazene-os no dicionário;
3. exiba os dados;
4. altere o preço;
5. adicione uma informação chamada categoria.

```python
print("=== Sistema de Cadastro de Biblioteca ===")

# 1. Leitura dos dados em variáveis temporárias
t = input("Qual o título do livro? ")
a = input("Quem é o autor? ")
ano_pub = int(input("Qual o ano de publicação? "))
valor = float(input("Qual o preço de venda? R$ "))

# 2. Armazenando no dicionário
obra_literaria = {
    "titulo": t,
    "autor": a,
    "ano": ano_pub,
    "preco": valor
}

# 3. Exibição inicial
print("\n-- Resumo da Obra --")
for chave, conteudo in obra_literaria.items():
    print(f" -> {chave.capitalize()}: {conteudo}")

# 4. Alteração de preço
print("\nHouve reajuste no valor!")
obra_literaria["preco"] = float(input("Informe o novo preço do livro: R$ "))

# 5. Adicionando nova informação (categoria)
obra_literaria["categoria"] = input("Classifique este livro (ex: Romance, Terror): ")

# Exibição final
print("\n-- Ficha Completa Atualizada --")
for chave, conteudo in obra_literaria.items():
    print(f"{chave.upper()}: {conteudo}")
```

## Atividade 4 — Dataclass
Crie uma dataclass chamada `Aluno` com nome, matrícula, nota1 e nota2.
Crie um método chamado `media()` que calcule a média das duas notas.

```python
from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    matricula: int
    nota1: float
    nota2: float

    # Método de cálculo dentro da classe
    def media(self) -> float:
        resultado = (self.nota1 + self.nota2) / 2
        return resultado

# Testando a criação do aluno
estudante = Aluno(
    nome="Carlos Eduardo",
    matricula=98765,
    nota1=7.5,
    nota2=8.5
)

print("=== Dados do Estudante ===")
print(f"Nome: {estudante.nome}")
print(f"Matrícula: {estudante.matricula}")
print(f"Média Final: {estudante.media():.1f}")
```

## Atividade 5 — Lista de objetos
Utilizando a dataclass `Aluno`, cadastre cinco alunos. Depois:
1. calcule a média de cada aluno;
2. exiba os alunos aprovados (média >= 7.0);
3. identifique o aluno com maior média.

```python
# Obs: Como estamos no Colab, a classe Aluno já foi importada na célula acima,
# mas vamos declará-la novamente para garantir que esta célula funcione sozinha.
from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    matricula: int
    nota1: float
    nota2: float

    def media(self) -> float:
        return (self.nota1 + self.nota2) / 2

# Lista para guardar os estudantes
turma = []

print("=== Registro da Turma ===")
# Lendo 5 alunos
for i in range(1, 6):
    print(f"\n- Estudante {i} -")
    n = input("Nome: ")
    m = int(input("Matrícula: "))
    n1 = float(input("1ª Nota: "))
    n2 = float(input("2ª Nota: "))

    # Adicionando o objeto criado à lista
    turma.append(Aluno(n, m, n1, n2))

print("\n=== Relatório de Aprovações ===")
# 1 e 2. Calculando a média e verificando aprovados
for estudante in turma:
    media_atual = estudante.media()
    if media_atual >= 7.0:
        print(f"[APROVADO] {estudante.nome} | Média: {media_atual:.2f}")

# 3. Aluno com maior média usando lambda
melhor_aluno = max(turma, key=lambda e: e.media())

print("\n=== Quadro de Honra ===")
print(f"A maior nota da turma foi de {melhor_aluno.nome}, com média {melhor_aluno.media():.2f}.")
```

## Desafio integrador — Sistema de estoque
Desenvolva um pequeno sistema de estoque criando uma dataclass `Produto`. O programa deverá permitir cadastrar, listar, buscar por código, calcular total em estoque, e exibir o item mais caro usando um Menu interativo.

```python
from dataclasses import dataclass

@dataclass
class Produto:
    codigo: int
    nome: str
    preco: float
    quantidade: int

    def valor_estoque(self) -> float:
        return self.preco * self.quantidade

# Inicializando o catálogo vazio e o controle do loop
catalogo = []
sistema_ativo = True

# O menu de repetição pedido no desafio
while sistema_ativo:
    print("\n" + "="*30)
    print("      GESTÃO DE ESTOQUE      ")
    print("="*30)
    print(" 1 - Cadastrar produto")
    print(" 2 - Listar produtos")
    print(" 3 - Buscar produto")
    print(" 4 - Valor total do estoque")
    print(" 5 - Produto mais caro")
    print(" 0 - Sair")
    print("="*30)

    opcao = input("Escolha a opção: ")

    if opcao == "1":
        print("\n>> CADASTRO DE PRODUTO <<")
        cod = int(input("Código numérico: "))
        desc = input("Nome do produto: ")
        valor = float(input("Preço: R$ "))
        qtd = int(input("Quantidade: "))

        catalogo.append(Produto(cod, desc, valor, qtd))
        print("-> Produto cadastrado com sucesso!")

    elif opcao == "2":
        print("\n>> CATÁLOGO ATUAL <<")
        if len(catalogo) == 0:
            print("Estoque vazio.")
        else:
            for item in catalogo:
                print(f"Cód: {item.codigo} | {item.nome} | R$ {item.preco:.2f} | Estoque: {item.quantidade} un.")

    elif opcao == "3":
        print("\n>> PESQUISA DE PRODUTO <<")
        codigo_alvo = int(input("Código para busca: "))
        achou = False

        for item in catalogo:
            if item.codigo == codigo_alvo:
                print(f"Encontrado: {item.nome} - R$ {item.preco:.2f} (Restam {item.quantidade})")
                achou = True

        if achou == False:
            print("-> Produto não localizado.")

    elif opcao == "4":
        print("\n>> RESUMO FINANCEIRO <<")
        patrimonio = 0.0
        for item in catalogo:
            patrimonio += item.valor_estoque()

        print(f"O valor total do estoque é: R$ {patrimonio:.2f}")

    elif opcao == "5":
        print("\n>> PRODUTO MAIS CARO <<")
        if len(catalogo) == 0:
            print("Não há produtos cadastrados.")
        else:
            item_topo = max(catalogo, key=lambda p: p.preco)
            print(f"O produto mais caro é: {item_topo.nome} (R$ {item_topo.preco:.2f}).")

    elif opcao == "0":
        print("\nEncerrando o sistema. Até logo!")
        sistema_ativo = False

    else:
        print("\nOpção inválida.")
```
