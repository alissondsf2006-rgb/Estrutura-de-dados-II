### Exercício 1 | Simulador "Desfazer"

**Objetivo:** Implemente uma Pilha para gerenciar ações em um editor de texto. Armazene cada ação (digitar, apagar, substituir). Ao acionar a função 'desfazer', o programa deve remover e reverter estritamente o último comando inserido.

```python
# A Pilha em Python é representada por uma lista simples
historico_comandos = []

def executar_acao(acao):
    historico_comandos.append(acao) # Insere no topo
    print(f"Ação executada: '{acao}' adicionada ao histórico.")

def desfazer_acao():
    if historico_comandos: # Verifica se a pilha não está vazia
        acao_desfeita = historico_comandos.pop() # Remove do topo
        print(f"Desfazendo a última ação: '{acao_desfeita}'")
    else:
        print("Nada para desfazer. O histórico está vazio.")

# Simulando o uso
print("--- Editor de Texto ---")
executar_acao("digitar 'Olá'")
executar_acao("substituir 'Olá' por 'Oi'")
executar_acao("apagar linha 1")

print("\n--- Pressionando Ctrl+Z ---")
desfazer_acao()
desfazer_acao()
```

### Exercício 2 | Sistema de Impressão
**Objetivo:** Implemente uma Fila para gerenciar um spooler de impressora. Cada novo documento é enfileirado. O sistema deve garantir que o documento impresso seja sempre o mais antigo aguardando na fila[cite: 1].

```python
from collections import deque

# A Fila é otimizada usando deque (Double-Ended Queue)
fila_impressao = deque()

def enviar_para_impressora(documento):
    fila_impressao.append(documento) # Insere no final
    print(f"Documento '{documento}' enfileirado para impressão.")

def processar_impressao():
    if fila_impressao:
        doc_impresso = fila_impressao.popleft() # Remove do início
        print(f"🖨️ Imprimindo agora: '{doc_impresso}'...")
    else:
        print("A fila de impressão está vazia.")

# Simulando o uso
print("--- Spooler da Impressora ---")
enviar_para_impressora("Relatorio_Mensal.pdf")
enviar_para_impressora("Boleto.pdf")
enviar_para_impressora("Foto_Ferias.png")

print("\n--- Processando Fila ---")
processar_impressao()
processar_impressao()
```

### Desafio Master | Triagem Hospitalar
**Cenário:** Criar um programa de gerenciamento para uma fila de atendimento médico.
*   **Fluxo Padrão:** Pacientes normais entram no final da fila.
*   **Fluxo de Exceção:** Pacientes prioritários (idosos, emergências) devem furar a fila e entrar diretamente no início do atendimento.

```python
from collections import deque

fila_hospital = deque()

def triagem(nome, prioritario=False):
    if prioritario:
        fila_hospital.appendleft(nome) # Fura a fila entrando pela esquerda (início)
        print(f"🚨 URGÊNCIA: '{nome}' inserido no início da fila!")
    else:
        fila_hospital.append(nome) # Entra normalmente pela direita (final)
        print(f"Atendimento normal: '{nome}' aguardando no final da fila.")

def chamar_paciente():
    if fila_hospital:
        proximo = fila_hospital.popleft()
        print(f"🩺 Médico chama: {proximo}")
    else:
        print("Nenhum paciente aguardando.")

# Simulando o uso
print("--- Recepção do Hospital ---")
triagem("João (Dor de cabeça)")
triagem("Maria (Exame de rotina)")
triagem("Sr. Carlos (Infartando, Idoso)", prioritario=True)
triagem("Ana (Gripe)")

print("\n--- Consultórios Abertos ---")
chamar_paciente() # Sr. Carlos deve ser o primeiro
chamar_paciente() # João é o próximo
```

### 01 | A Pilha Matemática
**Missão:** Crie uma pilha (LIFO) que armazene números inteiros. Escreva um algoritmo que esvazie a pilha iterativamente, calculando e retornando a soma total dos elementos removidos.

```python
pilha_numeros = [10, 25, 5, 40, 20] # Pilha pré-preenchida

def esvaziar_e_somar(pilha):
    soma_total = 0
    print(f"Pilha original: {pilha}")

    while pilha: # Enquanto houver elementos na pilha
        valor_removido = pilha.pop()
        soma_total += valor_removido
        print(f"Removido: {valor_removido} | Subtotal: {soma_total}")

    return soma_total

resultado = esvaziar_e_somar(pilha_numeros)
print(f"\nSoma total dos elementos retirados da pilha: {resultado}")
```

### 02 | O Call Center
**Missão:** Crie uma fila (FIFO) que armazene nomes de clientes. Simule um ciclo contínuo de atendimento: adicionar novos clientes à espera e "chamar" o próximo cliente da fila para atendimento.

```python
from collections import deque
import time

fila_espera = deque(["Cliente A", "Cliente B", "Cliente C"])

def atendimento_call_center():
    print(f"Fila inicial: {list(fila_espera)}\n")

    # Novo cliente ligando
    novo_cliente = "Cliente D"
    fila_espera.append(novo_cliente)
    print(f"📞 Nova ligação: {novo_cliente} entrou na fila.")

    # Atendendo
    while fila_espera:
        em_atendimento = fila_espera.popleft()
        print(f"🎧 Operador atendendo agora: {em_atendimento}...")
        time.sleep(1) # Simula o tempo de atendimento

atendimento_call_center()
print("\nTodos os clientes foram atendidos.")
```

### 01 | Pilha via Array Estático
**Missão:** Implemente uma pilha em C alocando um array de tamanho fixo. Desenvolva as funções `push()` e `pop()`, criando uma variável de controle manual para rastrear o índice exato do topo (`top`).

```python
#include <stdio.h>
#define TAMANHO_MAX 5

// Estrutura da Pilha
int pilha[TAMANHO_MAX];
int top = -1; // Variável de controle manual que rastreia o índice do topo

// Função para inserir na pilha
void push(int valor) {
    if (top == TAMANHO_MAX - 1) {
        printf("Erro: Pilha Cheia (Stack Overflow)\n");
    } else {
        top++;
        pilha[top] = valor;
        printf("Push: %d inserido no topo (indice %d)\n", valor, top);
    }
}

// Função para remover da pilha
void pop() {
    if (top == -1) {
        printf("Erro: Pilha Vazia (Stack Underflow)\n");
    } else {
        printf("Pop: %d removido do topo (indice %d)\n", pilha[top], top);
        top--;
    }
}

int main() {
    printf("--- Testando Pilha em C ---\n");
    push(10);
    push(20);
    push(30);
    pop();
    pop();

    return 0;
}
```

### 02 | Fila Circular
**Missão:** Otimize o uso de memória implementando uma fila sobre um array circular[cite: 1]. Desenvolva `enqueue()` e `dequeue()`, controlando independentemente os ponteiros de início (`front`) e fim (`rear`)[cite: 1].

```python
#include <stdio.h>
#define TAM_MAX 5

int fila[TAM_MAX];
int front = -1; // Início
int rear = -1;  // Fim

// Função para enfileirar
void enqueue(int valor) {
    // Verifica se está cheia usando aritmética modular (circularidade)
    if ((rear + 1) % TAM_MAX == front) {
        printf("Erro: Fila Circular Cheia!\n");
        return;
    }

    // Se estiver vazia, inicializa ambos no índice 0
    if (front == -1) {
        front = 0;
        rear = 0;
    } else {
        // Movimenta o final de forma circular
        rear = (rear + 1) % TAM_MAX;
    }

    fila[rear] = valor;
    printf("Enqueue: %d inserido na posicao %d\n", valor, rear);
}

// Função para desenfileirar
void dequeue() {
    if (front == -1) {
        printf("Erro: Fila Circular Vazia!\n");
        return;
    }

    printf("Dequeue: %d removido da posicao %d\n", fila[front], front);

    // Se era o último elemento, a fila volta a ficar vazia
    if (front == rear) {
        front = -1;
        rear = -1;
    } else {
        // Movimenta o início de forma circular
        front = (front + 1) % TAM_MAX;
    }
}

int main() {
    printf("--- Testando Fila Circular em C ---\n");
    enqueue(100);
    enqueue(200);
    enqueue(300);
    dequeue(); // Remove o 100
    enqueue(400); // Entra na posição seguinte

    return 0;
}
```
